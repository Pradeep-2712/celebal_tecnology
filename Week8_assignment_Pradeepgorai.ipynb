# Week 8 Assignment - Single Agent Pipeline Project

Build a Single-Agent Smart Assistant.

```python
import json
import re
```

## Calculator Tool

```python
def calculator(expression):
    allowed='0123456789+-*/().% '
    if any(ch not in allowed for ch in expression):
        raise ValueError('Invalid characters in expression.')
    return eval(expression, {'__builtins__': {}}, {})
```

## Keyword Extraction Tool

```python
def extract_keywords(text):
    stop={'the','is','a','an','and','or','to','of','in','on','for','with','from','this','that','it','are','was'}
    words=re.findall(r'[A-Za-z]+', text.lower())
    return [w for w in words if w not in stop]
```

## Agent

```python
def agent(query):
    print('='*50)
    print(f'[INFO] Received Query: {query}')
    try:
        query=query.strip()
        if not query:
            return {'type':'error','result':'Query cannot be empty.'}

        q=query.lower()

        if any(x in q for x in ['calculate','solve','compute']):
            print('[INFO] Routing -> Calculator Tool')
            expr=query
            for x in ['calculate','solve','compute']:
                expr=re.sub(x,'',expr,flags=re.IGNORECASE)
            return {'type':'calculation','result':calculator(expr.strip())}

        if any(x in q for x in ['keyword','keywords','extract keyword','extract keywords']):
            print('[INFO] Routing -> Keyword Tool')
            text=re.sub(r'extract\s+keywords?\s+from','',query,flags=re.IGNORECASE)
            text=re.sub(r'keywords?','',text,flags=re.IGNORECASE).strip()
            return {'type':'keywords','result':extract_keywords(text)}

        print('[INFO] Routing -> General Response')
        return {'type':'general','result':'I can perform calculations, extract keywords, and answer general queries.'}

    except Exception as e:
        print('[ERROR]',e)
        return {'type':'error','result':str(e)}
```

## Test Cases

```python
tests=[
'Calculate 25+15',
'Solve 100/4',
'Extract keywords from Artificial Intelligence is transforming healthcare rapidly',
'What is Machine Learning?',
'Calculate 5 //'
]
for t in tests:
    print('\n',json.dumps(agent(t),indent=4))
```

## Interactive Loop

```python
while True:
    q=input("Enter query (type 'exit' to stop): ")
    if q.lower()=='exit':
        print('Goodbye!')
        break
    print(json.dumps(agent(q),indent=4))
```
